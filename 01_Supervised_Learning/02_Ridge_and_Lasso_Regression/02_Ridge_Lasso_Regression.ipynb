Ridge Regression (L2 Regularization)
Theory
Ridge Regression is a regularization technique that reduces large model coefficients to prevent overfitting.

Mathematical formula:
\[
Loss = RSS + \alpha \sum \beta_j^2
\]
Where:
- RSS = Residual Sum of Squares.
- α = Regularization strength.
- β = Model coefficients.

alpha=1.0 controls the regularization strength.
- Smaller alpha = weaker regularization.
- Larger alpha = stronger regularization.
Ridge shrinks coefficients but generally does not make them exactly zero.

```python
from sklearn.linear_model import Ridge
```

```python
ridge = Ridge(alpha=1.0)

ridge.fit(X_train_scaled,y_train)
ridge_pred = (X_test_scaled)

print("Ridge Regression Predictions:")
print(ridge_pred)
```

Lasso Regression (L1 Regularization)
Theory
Lasso Regression is another regularization technique.

Unlike Ridge, Lasso can reduce some coefficients exactly to zero.

Mathematical formula:
\[
Loss = RSS + \alpha \sum |\beta_j|
\]

```python
from sklearn.linear_model import Lasso
```

```python
lasso = Lasso(alpha=100)
```

```python
lasso.fit(X_train_scaled,y_train)
lasso_pred = (X_test_scaled)

print("Lasso Regression Predictions:")
print(lasso_pred)
```

```python
results = []

for name, predictions in models.items():

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "MSE": mean_squared_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2 Score": r2_score(y_test, predictions)
    })

results_df = pd.DataFrame(results)

results_df
```
